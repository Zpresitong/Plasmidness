```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np

genus_name = 'Escherichia'
base_folder = f'/active-data/analysis_results/chr_pla/genus'
figure_fold = f'{base_folder}/figures'
os.chdir(figure_fold)

pic1 = mpimg.imread(f"{genus_name}_NMS_MS_link1.png")
pic2 = mpimg.imread(f"{genus_name}_NMS_MS_link_example1_filted.png")
pic3 = mpimg.imread(f"{genus_name}_sequence_exchange_example1.png")

h1, w1 = pic1.shape[0], pic1.shape[1]
h2, w2 = pic2.shape[0], pic2.shape[1]
h3, w3 = pic3.shape[0], pic3.shape[1]

total_width = max(w1, w2, w3)

def pad_to_width(img, target_width):
    h, w = img.shape[:2]
    if w >= target_width:
        return img
    pad_width = target_width - w
    return np.pad(img, ((0, 0), (0, pad_width), (0, 0)), mode='constant', constant_values=1)

pic1_pad = pad_to_width(pic1, total_width)
pic2_pad = pad_to_width(pic2, total_width)
pic3_pad = pad_to_width(pic3, total_width)

combined_img = np.vstack([pic1_pad, pic2_pad, pic3_pad])
total_height = h1 + h2 + h3

dpi = 1000
fig_width = total_width / dpi
fig_height = total_height / dpi

plt.figure(figsize=(fig_width, fig_height), dpi=dpi)
plt.imshow(combined_img)

plt.axis('off')
plt.subplots_adjust(left=0, right=1, top=1, bottom=0)

output_path = f"{genus_name}_tree_sequence_combined_3figs.png"
plt.savefig(output_path, dpi=dpi, bbox_inches='tight', pad_inches=0)
plt.show()
```
