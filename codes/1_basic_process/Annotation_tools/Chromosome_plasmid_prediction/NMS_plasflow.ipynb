```python
import pandas as pd
import ast
import os

def extract_name(x):
    try:
        genus_dict = ast.literal_eval(x)
        return genus_dict.get('name', None)
    except:
        return None

os.chdir('/active-data/analysis_results/chr_pla')
all_data = pd.read_csv('genome_chr-pla_statistics.csv')
all_data['genus_clean'] = all_data['genus'].apply(extract_name)
counts = all_data['genus_clean'].value_counts()
keep_genus = list(counts[counts >= 400].index)
print(keep_genus)
```

```python
for genus_name in keep_genus:
    print(genus_name)
    folder = '/active-data/analysis_results/chr_pla/genus/NMS_replicon/%s/' % genus_name
    data_dir = '/active-data/analysis_results/chr_pla/genus/statistics_records/%s/annotations' % genus_name
    os.makedirs(data_dir, exist_ok=True)
    
    os.chdir(folder)
    cmd = 'PlasFlow.py --input NMS_replicon_nucleotide_seq.fasta --output %s/PlasFlow_NMS_replicon_result.txt' % data_dir
    %time os.system(cmd)
    os.system('rm %s/PlasFlow_NMS_replicon_result.txt_chromosomes.fasta' % data_dir)
    os.system('rm %s/PlasFlow_NMS_replicon_result.txt_plasmids.fasta' % data_dir)
    os.system('rm %s/PlasFlow_NMS_replicon_result.txt_unclassified.fasta' % data_dir)
```
