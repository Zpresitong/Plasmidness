```python
import ast
import os
import pandas as pd
from tqdm import tqdm

def extract_name(x):
    try:
        genus_dict = ast.literal_eval(x)
        return genus_dict.get('name', None)
    except:
        return None

label = 'pident_90'

os.chdir('/active-data/analysis_results/chr_pla')
all_data = pd.read_csv('genome_chr-pla_statistics.csv')
all_data['genus_clean'] = all_data['genus'].apply(extract_name)
counts = all_data['genus_clean'].value_counts()
keep_genus = counts[counts >= 400].index.to_list()
print(keep_genus)
```

```python
# NMS_replicon_link

for genus_name in keep_genus:
    file_folder = f'/active-data/analysis_results/chr_pla/genus/cor-pla_fraction_records/{genus_name}'
    target_folder = f'/active-data/analysis_results/chr_pla/genus/statistics_records/{genus_name}'
    replicon_data = pd.read_csv(f'{target_folder}/replicon-plasmid_fraction-self_bitscore_statistics.csv')
    NMS_replicons = replicon_data[replicon_data['ms-label'] == 'NMS_replicon'].reset_index(drop=True)

    cut_value = 2/3
    link_count = 0
    NMS_links = []
    with tqdm(total = len(NMS_replicons), desc=f'{genus_name}({len(NMS_replicons)})', leave=True, ncols=100, unit='B', unit_scale=True) as pbar:
        for i in NMS_replicons.index:
            rep_name = NMS_replicons['accession'][i]
            acc_n, replicon = rep_name.split('-')[0], rep_name.split('-')[1]
            pbar.update(1)
            try:
                cover_csv = pd.read_csv(f'{file_folder}/{acc_n}/{replicon}_coverage.csv')
                cover_csv = cover_csv[(cover_csv['id'].isin(NMS_replicons['accession'])) & ~(cover_csv['id'] == rep_name) & (cover_csv[f'coverage-{label}'] >= cut_value)]
                cover_csv['source'] = rep_name
                cover_csv = cover_csv[['source', 'id', f'coverage-{label}']]
                cover_csv.columns = ['source', 'target', 'coverage']
                NMS_links.append(cover_csv)
                link_count += len(cover_csv)
            except:
                continue
    NMS_link_pd = pd.concat(NMS_links, ignore_index = True)
    NMS_link_pd.to_csv(f'{target_folder}/{genus_name}_NMS_replicon_link.csv', index=False)    
    print(genus_name, link_count)
```

```python
# typical_chromosome_link_with_sub
def safe_literal_eval(x):
    if pd.isna(x) or x == '':
        return []
    try:
        return ast.literal_eval(x)
    except:
        return []

def merge_intervals(intervals):
    if not intervals:
        return []
    
    parsed = []
    for s in intervals:
        try:
            start, end = map(int, s.split('-'))
            # 确保 start <= end
            if start > end:
                start, end = end, start
            parsed.append((start, end))
        except:
            continue
    
    parsed.sort()
    
    merged = [parsed[0]]
    for current in parsed[1:]:
        last = merged[-1]
        if current[0] <= last[1] + 1:
            new_start = last[0]
            new_end = max(last[1], current[1])
            merged[-1] = (new_start, new_end)
        else:
            merged.append(current)
    
    return [f"{s}-{e}" for s, e in merged]

def calculate_coverage(merged_range, rep_len):
    if not merged_range:
        return 0
    
    total_bp = 0
    for interval in merged_range:
        start, end = map(int, interval.split('-'))
        total_bp += end - start + 1
    
    coverage = total_bp / rep_len
    return coverage

for genus_name in keep_genus:
    file_folder = f'/active-data/analysis_results/chr_pla/genus/cor-pla_fraction_records/{genus_name}'
    target_folder = f'/active-data/analysis_results/chr_pla/genus/statistics_records/{genus_name}'
    replicon_data = pd.read_csv(f'{target_folder}/replicon-plasmid_fraction-self_bitscore_statistics.csv')
    ty_ch_replicons = replicon_data[replicon_data[f'category-{label}'] == 'typical chromosome'].reset_index(drop=True)
    sub_replicons = replicon_data[replicon_data[f'category-{label}'] != 'typical chromosome'].reset_index(drop=True)

    cut_value = 2/3
    link_count = 0
    sub_links = []
    with tqdm(total = len(sub_replicons), desc=f'{genus_name}({len(sub_replicons)})', leave=True, ncols=100, unit='B', unit_scale=True) as pbar:
        for i in sub_replicons.index:
            rep_name = sub_replicons['accession'][i]
            rep_len = sub_replicons['size'][i]
            acc_n, replicon = rep_name.split('-')[0], rep_name.split('-')[1]
            pbar.update(1)
            try:
                cover_csv = pd.read_csv(f'{file_folder}/{acc_n}/{replicon}_coverage.csv')
                cover_csv = cover_csv[(cover_csv['id'].isin(ty_ch_replicons['accession']))]
                cover_csv['acc'] = cover_csv['id'].str.split('-').str[0]
                cover_csv['range_list'] = cover_csv[f'range-{label}'].apply(safe_literal_eval)
                new_df = cover_csv.groupby('acc').agg(
                    all_ranges=('range_list', lambda x: [item for sublist in x for item in sublist])
                ).reset_index()
                new_df['merged_range'] = new_df['all_ranges'].apply(merge_intervals)
                new_df['coverage'] = new_df['merged_range'].apply(lambda x: calculate_coverage(x, rep_len))
                new_df = new_df[new_df['coverage'] >= cut_value]
                new_df['source'] = rep_name
                new_df = new_df[['source', 'acc', 'coverage']]
                new_df.columns = ['source', 'target', 'coverage']
                if not new_df.empty:
                    sub_links.append(new_df)
                link_count += len(new_df)
            except:
                continue
    NMS_link_pd = pd.concat(sub_links, ignore_index = True)
    NMS_link_pd.to_csv(f'{target_folder}/{genus_name}_typical_chromosome_sub_replicon_link.csv', index=False)    
    print(genus_name, link_count)
```
