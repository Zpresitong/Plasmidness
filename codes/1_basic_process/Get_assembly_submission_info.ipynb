```python
import os
import jsonlines
import pandas as pd

os.chdir('/active-data/genomes/bacteria_complete_annotationRefSeq-20250807/data')
acc_info = []
with open('assembly_data_report.jsonl', 'r') as file:
    for line in jsonlines.Reader(file):
        accession = line.get("accession", "")

        submission_date = None
        assembly_info = line.get("assemblyInfo", {})
        biosample = assembly_info.get("biosample", {})
        submission_date = biosample.get("submissionDate", "")

        submitter = assembly_info.get("submitter", "")
        seqtech = assembly_info.get("sequencingTech", "")

        acc_info.append({
            "accession": accession,
            "submissionDate": submission_date,
            "submitter": submitter,
            "sequencingTech": seqtech
        })

df = pd.DataFrame(acc_info)
os.chdir('/active-data/genomes/bacteria_complete_annotationRefSeq-20250807')
df.to_csv("assembly_submission_info.tsv", sep='\t', index=False)
```
